# Point Class

```python
class Point:
    def __init__(self,x,y):   # Construnctor takes 2 mandatory parameters x and y
        self.abscissa = x   # x coordinate is called abscissa
        self.ordinate = y   # y coordinate is called ordinate

    def __str__(self):
        return '({},{})'.format(self.abscissa,self.ordinate)

    def eucledian_distance(self,other):
        # Formula √((x2-x1)^2+(y2-y2)^2)
        return (((other.abscissa-self.abscissa)**2)+((other.ordinate-self.ordinate)**2))**0.5
    def distance_from_origin(self):
        return self.eucledian_distance(Point(0,0))
        # return (((self.abscissa)**2)+((self.ordinate)**2))**0.5
```

## Line Class

```python
class Line:
    def __init__(self, A, B, C):
        self.A = A
        self.B = B
        self.C = C
    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)
    def distance_from_point(self,Point):
        # Formula |Ax1+By1+C|/√(A^2+B^2)
        return abs((self.A*Point.abscissa)+(self.B*Point.ordinate)+self.C)/((self.A**2)+(self.B)**2)**0.5
    def is_point_on_line(self,Point):
        if (self.A*Point.abscissa)+(self.B*Point.ordinate)+self.C == 0:
            return True
        return False
    def point_of_intersection(self,other):
        den = self.A*other.B-self.B*other.A
        if den == 0:
            return "Given Two lines Never intersect"
        abscissa = (self.B*other.C - other.B*self.C)/den
        ordinate = (self.C*other.A - other.C*self.A)/den
        return Point(abscissa,ordinate)
```

## Verification
