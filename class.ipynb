# Functions Vs Methods
### Function is dfferent from method
### Functions in class are called methods
### Functions that exist idependently and can be executed independently are called functions

### li = [1,2,3]
### len(li)  # Len is a function because it is outside list class
### li.append(5) # append is a method as it is inside list class

```python
class Atm():
    def __init__(self):  
        self.pin = ''
        self.balance = 0
        # self.main()
        print("Welcome to the ATM!")
        self.menu()
    def menu(self):
        user_input = input("Hi How can I help you? \n" \
        "Press 1 to set your pin \n" \
        "Press 2 to chnage your pin \n" \
        "Press 3 to check your balance \n" \
        "Press 4 to withdraw \n" \
        "Press 5 to deposit \n" \
        "Press anything else to exit")
        if user_input == '1':
            self.set_pin()
        elif user_input == '2':
            # self.change_pin()
            self.change_pin()
        elif user_input == '3':
            # self.check_balance()
            self.check_balance()
        elif user_input == '4':
            # self.withdraw()
            self.withdraw()
        elif user_input == '5':            # self.deposit()
            self.deposit()
        else:
            print("Thank you for using our ATM. Goodbye!")
    def set_pin(self):
        user_pin = int(input("Please enter a new 4-digit PIN: "))
        if len(str(user_pin)) == 4:
            self.pin = user_pin
            print("Your PIN has been set successfully.")
        self.menu()
    def change_pin(self):
        old_pin = int(input("Please enter your current PIN: "))
        if old_pin == self.pin:
            new_pin = int(input("Please enter a new 4-digit PIN: "))
            if len(str(new_pin)) == 4:
                self.pin = new_pin
                print("Your PIN has been changed successfully.")
        else:
            print("Incorrect PIN. Please try again.")
        self.menu()
    def check_balance(self):
        user_pin = int(input("Please enter your PIN: "))
        if user_pin == self.pin:
            print(f"Your current balance is: ${self.balance}")
        else:
            print("Incorrect PIN. Please try again.")
        self.menu()
    def withdraw(self):
        user_pin = int(input("Please enter your PIN: "))
        if user_pin == self.pin:
            amount = int(input("Please enter the amount to withdraw: "))
            if amount <= self.balance:
                self.balance -= amount
                print(f"You have withdrawn ${amount}. Your new balance is: ${self.balance}")
            else:
                print("Insufficient funds. Please try again.")
        else:
            print("Incorrect PIN. Please try again.")
        self.menu()
    def deposit(self):
        user_pin = int(input("Please enter your PIN: "))
        if user_pin == self.pin:
            amount = int(input("Please enter the amount to deposit: "))
            self.balance += amount
            print(f"You have deposited ${amount}. Your new balance is: ${self.balance}")
        else:
            print("Incorrect PIN. Please try again.")
        self.menu()
```

```python
manoj = Atm()
```

## Parameterized Constructor

### A constructor that takes additional arguments besides the standard self reference. These arguments are used to initialize the object's attributes or perform other operations when the object is created.

```python
class Family:
    # Constructor - parameterized
    members = 5

    def __init__(self, count):
        print("This is a parameterized constructor")
        self.members = count

    def show(self):
        print("Number of members is", self.members)

family_object = Family(10)
family_object.show()
```

## Non-Parameterized Constructor
### A constructor that does not accept any arguments except the mandatory self. This type of constructor is used for initializing class members with default values or performing standard initialization tasks that do not require external inputs.

```python
class Fruits:
    favourite = "Apple"

    # Non-parameterized constructor
    def __init__(self):
        self.favourite = "Orange"

    # A method to display the favorite fruit
    def show(self):
        print(self.favourite)

# Creating an object of the class
obj = Fruits()

# Calling the instance method using the object
obj.show()
```

## Default Constructor
### When a class is defined without an explicit constructor in Python, Python automatically provides a default constructor. This default constructor is a basic, non-parameterized constructor that performs no specific task or initialization beyond the basic object creation.

```python
class Assignments:
    check = "not done"

    # A method to check assignment status
    def is_done(self):
        print(self.check)

# Creating an object of the class
obj = Assignments()

# Calling the instance method using the object
obj.is_done()
```

# Self
### Golden Rule of OOP
### Only an object of a class can access or call all the data/variables or methods in that class 
### Means not even a method in class cannot call another method in same class or access a variable in same class
### This problem is solved by self key word

```python
class ExplainSelf():
    def __init__(self):
        self.name = "Manoj"
        print(id(self))
```

```python
manoj = ExplainSelf()
```

```python
print(id(manoj))
```

### If you see above lines of code, self and the object are stored in same address, which means the self is nothing but the object itself, that is why it is able to call the method in the class and helps in calling methods or accessing variables in the same class

```python
class ExplainSelf():
    def __init__(self):
        self.name = "Manoj"
        #abc()  #Throws an error if we try to access method without self as only an object can access variables or methods present inside a class and not even the class itself.
        self.defg() #This is the correct way to access a method inside a class because we are using self which is an instance of the class to access the method.
    def abc(self):
        print("This is a method inside the class.")
    def defg(self):
        print("This is another method inside the class.")
```

### Whenever we call a method of a class with object, the object itself will be passed as parameter to the method

```python
class Explainer():
    def __init__(self):
        self.explanation = "This is a simple ATM class that allows users to set and change their PIN, check their balance, withdraw and deposit money. The class uses a menu system to interact with the user and performs the necessary operations based on the user's input."
        print(self.explanation)
    def abc(self):
        print("This is a method inside the class.")
    def defg():
        print("This is another method inside the class.")
```

```python
obj = Explainer()
```

```python
obj.abc()  #This works because we used self to access the method inside the class. If we try to access the method without self it will throw an error because only an object can access variables or methods present inside a class and not even the class itself.
obj.defg()  #This will throw an error because we did not use self while defining the method inside the class and when object tries to use method the object itself will be passed as an argument to the method and since we did not use self while defining the method it will throw an error because the method is expecting no arguments but we are passing one argument which is the object itself.
```

### In simple words self is nothing but the current object which we are working with
### And why we should keep self inside a class is because OOP has golden rule that only an object can access variables/attributes and methods inside a class, so you need an object of the class inside the class which self will do (self is just a name, you can use any other name unlike __init__)
